```python
# 🟢 Cell 1: Basic Positive Slicing (Cutting the front/middle)
# --- Level 1: Basic Forward Slicing ---
word = "PYTHON"

# Grab from index 0 up to (but not including) index 2
# This will pick indices 0 and 1 -> 'P' and 'Y'
slice_front = word[0:2]

# Grab from index 2 up to (but not including) index 5
# This will pick indices 2, 3, and 4 -> 'T', 'H', and 'O'
slice_middle = word[2:5]

print("Front Slice [0:2]:", slice_front)    # Expected Output: PY
print("Middle Slice [2:5]:", slice_middle)  # Expected Output: THO
```

```python
# 🟡 Cell 2: The Missing Shortcuts (Leaving blanks)
# --- Level 2: Slicing Shortcuts ---
text = "JupyterNotebook"

# Shortcut 1: Leave 'start' blank to begin from the very start (index 0)
# Goes from index 0 up to 7 -> 'Jupyter'
first_word = text[:7]

# Shortcut 2: Leave 'stop' blank to go all the way to the very end
# Starts at index 7 and grabs everything left -> 'Notebook'
second_word = text[7:]

# Shortcut 3: Leave both blank to clone the entire string
full_copy = text[:]

print("First Word [:7]:", first_word)     # Expected Output: Jupyter
print("Second Word [7:]:", second_word)   # Expected Output: Notebook
print("Full Copy [:]:", full_copy)        # Expected Output: JupyterNotebook
```

```python
# 🟠 Cell 3: Negative Slicing (Cutting from the back)
# --- Level 3: Negative Index Slicing ---
filename = "report_2026.pdf"

# Grab the last 3 characters of the string
# Goes from index -3 to the very end
extension = filename[-3:]

# Grab the year section using negative numbers
# Starts at -8 ('2') and stops right before -4 ('.')
year = filename[-8:-4]

print("File Extension [-3:]:", extension) # Expected Output: pdf
print("Extracted Year [-8:-4]:", year)    # Expected Output: 2026
```

```python
# 🔴 Cell 4: Stepping and Skipping (The Step Value)
# --- Level 4: Using Steps ---
numbers = "0123456789"

# Take every 2nd character from the entire string
# Starts at 0, skips 1, takes 2, skips 3, takes 4...
evens = numbers[::2]

# Start at index 1, stop at index 8, step by 3
# Takes index 1 ('1'), adds 3 to get index 4 ('4'), adds 3 to get index 7 ('7')
skips = numbers[1:8:3]

print("Every 2nd number [::2]:", evens)    # Expected Output: 02468
print("Custom step gap [1:8:3]:", skips)   # Expected Output: 147
```

```python
# 🔥 Cell 5: The Advanced String Reversal Trick
# --- Level 5: Reversing a String ---
secret_message = "stressed"

# Step by -1 means move backward one character at a time
reversed_message = secret_message[::-1]

print("Original:", secret_message)
print("Reversed [::-1]:", reversed_message) # Expected Output: desserts
```

```python
# --- Level 4: BODMAS Challenge ---
combined_expr = 20 - 3 * 2 ** 2 + 15 // 4

# Step-by-step breakdown done by Python:
# Step 1 (Power): 2 ** 2 becomes 4       -> (20 - 3 * 4 + 15 // 4)
# Step 2 (Mul):   3 * 4 becomes 12       -> (20 - 12 + 15 // 4)
# Step 3 (Floor): 15 // 4 becomes 3      -> (20 - 12 + 3)
# Step 4 (Sub):   20 - 12 becomes 8      -> (8 + 3)
# Step 5 (Add):   8 + 3 becomes 11

print("Combined Expression Result:", combined_expr)   # Expected: 11
```

```python
# --- Level 5: The Ultimate Mix ---
ultimate_expr = (5 + 5) * 2 ** 3 / 4 + 7 // 2

# Step-by-step breakdown done by Python:
# Step 1 (Brackets):  (5 + 5) becomes 10               -> (10 * 2 ** 3 / 4 + 7 // 2)
# Step 2 (Power):     2 ** 3 becomes 8                 -> (10 * 8 / 4 + 7 // 2)
# Step 3 (Mul):       10 * 8 becomes 80                -> (80 / 4 + 7 // 2)
# Step 4 (Stand Div): 80 / 4 becomes 20.0 (FLOAT!)     -> (20.0 + 7 // 2)
# Step 5 (Floor Div): 7 // 2 becomes 3                 -> (20.0 + 3)
# Step 6 (Add):       20.0 + 3 converts to 20.0 + 3.0  -> 23.0

print("Ultimate Expression Result:", ultimate_expr)   # Expected: 23.0
```

# 3)Comparison Operator

```python
# 🟢 Cell 1: Equal to (==) and Not Equal to (!=)
# --- Level 1: Equality Checks ---
my_age = 20
your_age = 25

# Is my_age equal to your_age?
is_equal = (my_age == your_age)

# Is my_age NOT equal to your_age?
is_different = (my_age != your_age)

print("Are our ages equal? (==):", is_equal)         # Expected Output: False
print("Are our ages different? (!=):", is_different) # Expected Output: True
```

```python
# 🟡 Cell 2: Greater Than (>) and Less Than (<)
# --- Level 2: Basic Size Checks ---
wallet_a = 50
wallet_b = 100

print("Is wallet_a greater than wallet_b? (>):", wallet_a > wallet_b) # Expected Output: False
print("Is wallet_a less than wallet_b? (<):", wallet_a < wallet_b)    # Expected Output: True
```

```python
# 🟠 Cell 3: Greater Than or Equal To (>=) and Less Than or Equal To (<=)
# --- Level 3: Inclusive Boundary Checks ---
passing_score = 40
my_score = 40

# Checks if my_score is greater than OR exactly equal to passing_score
print("Did I pass? (>=):", my_score >= passing_score) # Expected Output: True

# Let's check another example with less than or equal to
max_capacity = 50
current_guests = 45
print("Are we under or at capacity? (<=):", current_guests <= max_capacity) # Expected Output: True
```

```python
# 🔴 Cell 4: Comparing Different Data Types (The Trick Cell)
# --- Level 4: Mixing Data Types ---

# 1. Comparing an Integer to a Float with the same math value
print("Does 5 == 5.0?:", 5 == 5.0)   # Expected Output: True (Python is smart enough to match values)

# 2. Comparing text strings (Strings are case-sensitive!)
print("Does 'apple' == 'Apple'?:", "apple" == "Apple") # Expected Output: False

# 3. Comparing text numbers with real numbers
print("Does '10' == 10?:", "10" == 10) # Expected Output: False (A string of text is never equal to a number!)
```

```python
# --- Level 1: The 'and' Operator ---
has_ticket = True
is_wearing_shoes = True
is_under_age = False

# Condition: To enter, you need a ticket AND you must wear shoes
can_enter = has_ticket and is_wearing_shoes
print("Can enter the venue?:", can_enter)        # Expected Output: True

# Condition: You need a ticket AND you must be under age
gets_child_discount = has_ticket and is_under_age
print("Gets child discount?:", gets_child_discount) # Expected Output: False
```

```python
# 🔴 Cell 4: Combining Logical Operators (BODMAS for Logic)
# --- Level 4: The Combined Mix ---
has_membership = False
has_coupon = True
is_holiday = False

# Scenario: You get a discount if you have a membership OR a coupon, AND it is NOT a holiday.
# Let's see how Python reads this left-to-right with brackets forcing the order.
eligible_for_discount = (has_membership or has_coupon) and (not is_holiday)

# Step-by-step breakdown:
# Step 1: (has_membership or has_coupon) -> (False or True) -> True
# Step 2: (not is_holiday)               -> (not False)      -> True
# Step 3: True and True                  -> True

print("Is eligible for a discount?:", eligible_for_discount) # Expected Output: True
```

# 6)Identity Operators

```python
# 🟢 Cell 1: Value vs. Identity (The Clone Example)
# --- Level 1: Same Value vs Same Memory ---
list_a = [1, 2, 3]
list_b = [1, 2, 3]  # This creates a brand new list that happens to look like list_a

# Check values using ==
print("Do they have the same values? (==):", list_a == list_b)  # Expected Output: True

# Check identity (memory location) using 'is'
print("Are they the exact same object? (is):", list_a is list_b) # Expected Output: False
```

```python
# 🟡 Cell 2: Creating a True Alias (Pointing to the Same Memory)
# --- Level 2: Pointing to the Same Object ---
list_x = ["apple", "banana"]
list_y = list_x  # list_y now points to the exact same list in memory!

print("Are list_x and list_y the same object? (is):", list_x is list_y) # Expected Output: True

# Proof: If we modify list_y, list_x changes too!
list_y.append("orange")
print("Look at list_x now:", list_x)  # Output will include 'orange'
```

```python
# 🟠 Cell 3: The is not Operator
# --- Level 3: Checking for Different Objects ---
player_one_bag = ["sword", "shield"]
player_two_bag = ["sword", "shield"]

# Even though they have identical items, they are separate player inventories
is_separate_inventory = player_one_bag is not player_two_bag

print("Are these separate bags in memory?:", is_separate_inventory) # Expected Output: True
```

```python
# 🔴 Cell 4: Python's Integer Caching Trick (The Integer Exception)
# --- Level 4: Python's Hidden Number Trick ---

# Small numbers are recycled automatically by Python
num1 = 100
num2 = 100
print("Small numbers - Are they identical? (is):", num1 is num2) # Expected Output: True

# Large numbers are treated as separate objects
num3 = 999999
num4 = 999999
print("Large numbers - Are they identical? (is):", num3 is num4) # Expected Output: False
```

```python

#🟢 Cell 1: Changing the Text Case (.upper(), .lower(), .title())# --- Level 1: Changing Case ---
message = "learn python from scratch"

# 1. Convert everything to UPPERCASE
upper_msg = message.upper()

# 2. Convert everything to lowercase
lower_msg = upper_msg.lower()

# 3. Capitalize the first letter of every word (Title Case)
title_msg = message.title()


print("Original    :", message)
print("Uppercase   :", upper_msg)  # Expected: LEARN PYTHON FROM SCRATCH
print("Lowercase   :", lower_msg)  # Expected: learn python from scratch
print("Title Case  :", title_msg)  # Expected: Learn Python From Scratch
```

```python
# 🟠 Cell 3: Find and Replace (.replace(), .count())
# --- Level 3: Search and Replace ---
text = "I love Java. Java is great!"

# Replace all instances of 'Java' with 'Python'
new_text = text.replace("Java", "Python")

# Count how many times the word 'Python' appears in the new text
python_count = new_text.count("Python")

print("Replaced Text:", new_text)      # Expected: I love Python. Python is great!
print("Python count :", python_count)  # Expected: 2
```
