```python
# Remove any currency symbols (like $ or ₹), commas, or extra spaces, then convert to float
df["Revenue"] = pd.to_numeric(
    df["Revenue"].str.replace(r"[^\d.-]", "", regex=True), 
    errors="coerce"
)

# Now calculate Profit_Amount
df["Profit_Amount"] = df["Revenue"] * (df["Profit_Margin"] / 100)
```

```python
# Calculate actual profit earned per transaction: Revenue * (Profit_Margin / 100)
df["Profit_Amount"] = (
    df["Revenue"] *
    df["Profit_Margin"] / 100
)
```

```python
# Clean any commas or spaces and convert both columns to numbers
df["Quantity_Available"] = pd.to_numeric(
    df["Quantity_Available"].astype(str).str.replace(r"[^\d.-]", "", regex=True), 
    errors="coerce"
)

df["Units_Sold"] = pd.to_numeric(
    df["Units_Sold"].astype(str).str.replace(r"[^\d.-]", "", regex=True), 
    errors="coerce"
)

# Now subtract Units_Sold from Quantity_Available
df["Stock_Difference"] = df["Quantity_Available"] - df["Units_Sold"]
```

```python
# Calculate remaining inventory balance by subtracting Units_Sold from Quantity_Available
df["Stock_Difference"] = (
    df["Quantity_Available"] -
    df["Units_Sold"]
)
```

# Understand Data Distribution and Skewness

```python
# Understanding Skewness
```

```python
# Import Pandas for statistical skewness calculations and NumPy for numerical array handling
import pandas as pd
import numpy as np
```

```python
# Load the retail dataset (5,171 rows x 47 columns)
# Note: Use your cleaned DataFrame from Lab 6 (where Revenue and Units_Sold are float64)
df = pd.read_csv(r'/mnt/c/Users/navjyot/Downloads/Retail_Dataset_Cleaned.csv')
print("Dataset Loaded Successfully")
```

```python
# Display the first 5 records (rows 0 to 4) to inspect initial store transactions
df.head()
```

```python
# Verify column data types (confirm Revenue and Units_Sold are float64 before computing skewness)
df.info()
```

```python
# List all 47 column names (Store_ID, Category, Product_Name, Units_Sold, Revenue, etc.)
df.columns
```

```python
# Inspect the 'Revenue' Series across all 5,171 retail transactions
df["Revenue"]
```

```python
# Calculate Fisher-Pearson skewness coefficient for Revenue (0 = symmetric, >0 = right-skewed, <0 = left-skewed)
revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

```python
# Classify the direction of Revenue asymmetry based on the computed skewness value
if revenue_skew > 0:
    print("Positive Skew")
elif revenue_skew < 0:
    print("Negative Skew")
else:
    print("Symmetric Distribution")
```

```python
# Calculate skewness for Units_Sold to determine whether order quantities cluster at low or high volumes
units_skew = df["Units_Sold"].skew()
print(units_skew)
```
