# Clean and Prepare Dataset for Analysis

```python
# Fix Data Types
```

```python
# Import Pandas for tabular data cleaning and NumPy for numerical handling
import pandas as pd
import numpy as np
```

```python
# Load the retail dataset (5,171 rows x 47 columns) into a Pandas DataFrame
# Note: Use '/content/Retail_Dataset_Modified.csv' on Colab or r'/mnt/c/Users/navjyot/Downloads/Retail_Dataset_Modified.csv' locally
df = pd.read_csv(r'/mnt/c/Users/navjyot/Downloads/Retail_Dataset_Modified.csv')
print("Dataset Loaded Successfully")
```

```python
# Display the first 5 records to inspect sample values before type conversion
df.head()
```

```python
# Inspect current data types across all 47 columns (identifies string-encoded numeric columns)
print("Data Types:")
print(df.dtypes)
```

```python
# Check non-null counts and memory usage (~3.0 MB) prior to cleaning
df.info()
```

```python
# Convert 'Quantity_Available' from string/object to numeric (float64); invalid text becomes NaN
df['Quantity_Available'] = pd.to_numeric(
    df['Quantity_Available'],
    errors='coerce'
)
print(df['Quantity_Available'].dtypes)
```

```python
# Convert 'Units_Sold' from string/object to numeric (float64) so sales volume math works properly
df['Units_Sold'] = pd.to_numeric(
    df['Units_Sold'],
    errors='coerce'
)
print(df['Units_Sold'].dtypes)
```

```python
# Convert 'Revenue' from string/object to numeric (float64) to fix string concatenation issues in sums
df['Revenue'] = pd.to_numeric(
    df['Revenue'],
    errors='coerce'
)
print(df['Revenue'].dtypes)
```

```python
# Convert 'Shipping_Cost' from string/object to numeric (float64) for logistics cost analysis
df['Shipping_Cost'] = pd.to_numeric(
    df['Shipping_Cost'],
    errors='coerce'
)
print(df['Shipping_Cost'].dtypes)
```

```python
# Verify that all four target columns are now properly typed as float64
print(df[['Quantity_Available',
          'Units_Sold',
          'Revenue',
          'Shipping_Cost']].dtypes)
```

```python
# Handle Missing Values
```

```python
# Audit missing (NaN) values across all 47 columns (including newly coerced NaNs from Task 1)
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Filter and display only the columns that currently contain at least one missing value
missing_cols = df.isnull().sum()
print(missing_cols[missing_cols > 0])
```

```python
# Impute missing categorical 'City' values (initially 62 missing) using the most frequent city (Mode)
# Note: Direct assignment (df['col'] = df['col'].fillna(...)) is recommended over inplace=True in Pandas 3.x
df['City'] = df['City'].fillna(df['City'].mode())
```

```python
# Impute missing categorical 'Product_Name' values (initially 73 missing) using the Mode
df['Product_Name'] = df['Product_Name'].fillna(df['Product_Name'].mode())
```

```python
# Impute missing numerical 'Quantity_Available' values (created during coercion) using column Mean
df['Quantity_Available'] = df['Quantity_Available'].fillna(df['Quantity_Available'].mean())
```

```python
# Impute missing numerical 'Units_Sold' values (created during coercion) using column Mean
df['Units_Sold'] = df['Units_Sold'].fillna(df['Units_Sold'].mean())
```

```python
# Impute missing numerical 'Revenue' values (72 initial + any coerced text values) using column Mean
df['Revenue'] = df['Revenue'].fillna(df['Revenue'].mean())
```

```python
# Impute missing categorical 'Supplier' values (initially 68 missing) using the Mode
df['Supplier'] = df['Supplier'].fillna(df['Supplier'].mode())
```

```python
# Impute missing numerical 'Customer_Satisfaction' scores (initially 77 missing) using column Mean
df['Customer_Satisfaction'] = df['Customer_Satisfaction'].fillna(df['Customer_Satisfaction'].mean())
```

```python
# Impute missing numerical 'Delivery_Time' values (initially 60 missing) using column Mean
df['Delivery_Time'] = df['Delivery_Time'].fillna(df['Delivery_Time'].mean())
```

```python
# Impute missing numerical 'Shipping_Cost' values (created during coercion) using column Mean
df['Shipping_Cost'] = df['Shipping_Cost'].fillna(df['Shipping_Cost'].mean())
```

```python
# Confirm that all 47 columns now report 0 missing values
print(df.isnull().sum())
```

```python
# Re-verify DataFrame structure, non-null counts (all 5,171 non-null), and updated dtypes
df.info()
```
