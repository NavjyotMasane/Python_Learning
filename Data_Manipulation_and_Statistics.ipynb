# Apply NumPy Indexing, Slicing & Matrix Operations for Data Analysis

```python
# Load Dataset Using Pandas
```

```python
# Import the Pandas and numpy library and give it the alias 'pd' and 'np' for easy use
```

```python
import pandas as pd
import numpy as np
```

```python
# Load the retail CSV dataset into a Pandas DataFrame named df
df = pd.read_csv(r'/mnt/c/Users/navjyot/Downloads/Retail_Dataset_Modified.csv')  

print("Dataset Load Successfully")
```

```python
# Apply Indexing and Slicing
```

```python
# Extract the 'Revenue' column from the DataFrame and convert it into a 1D NumPy array
revenue_array = np.array(df["Revenue"])

print(revenue_array)
```

```python
# Access and display the first element of the revenue array using zero-based indexing
print("First Revenue Value:")
print(revenue_array[0])
```

```python
# Access and display the last element of the revenue array using negative indexing
print("Last Revenue Value:")
print(revenue_array[-1])
```

```python
# Slice and display the first five elements (indices 0 through 4) of the revenue array
print("First Five Revenue Values:")
print(revenue_array[0:5])
```

```python
# Slice and display elements from index 10 up to index 20 (inclusive) of the revenue array
print("Revenue Values from Index 10 to 20:")
print(revenue_array[10:21])
```

```python
# Select 'Units_Sold' and 'Profit_Margin' columns and convert them into a 2D NumPy array
sales_array = np.array(df[["Units_Sold", "Profit_Margin"]])

print(sales_array)
```

```python
# Access and display the entire first row (all columns at index 0) of the 2D sales array
print("First Row:")
print(sales_array[0])
```

```python
# Slice all rows (:) from column index 1 to extract the entire Profit_Margin column
print("Profit Margin Column:")
print(sales_array[:, 1])
```

```python
# Perform Operations
```

```python
# Slice the first 10 elements (indices 0 through 9) from revenue_array into a new subset
first_ten = revenue_array[0:10]
print(first_ten)
```

```python
# Calculate and display the total sum of the first 10 revenue values using np.sum()
print("Total Revenue:")
print(np.sum(first_ten))
```

```python
# Identify and display the maximum value in the first_ten revenue subset using np.max()
print("Maximum Revenue:")
print(np.max(first_ten))
```

```python
# Identify and display the minimum value in the first_ten revenue subset using np.min()
print("Lowest Revenue:")
print(np.min(first_ten))
```

```python
# Extract the 'Profit_Margin' column from the DataFrame and convert it into a 1D NumPy array
profit_array = np.array(df["Profit_Margin"])

print(profit_array)
```

```python
# Compute and display the arithmetic mean of the Profit_Margin array using np.mean()
print("Average Profit Margin:")
print(np.mean(profit_array))
```

```python
# Compute and display the median (middle value) of the Profit_Margin array using np.median()
print("Median Profit Margin:")
print(np.median(profit_array))
```

```python
# Identify and display the minimum and maximum profit margin values to establish the data range
print("Minimum Profit Margin:")
print(np.min(profit_array))

print("Maximum Profit Margin:")
print(np.max(profit_array))
```

```python
# Compute and display the cumulative sum of all values in the Profit_Margin array
print("Total Profit Margin:")
print(np.sum(profit_array))
```

```python
# Compute and display the standard deviation of the Profit_Margin array to measure data dispersion
print("Standard Deviation:")
print(np.std(profit_array))
```

```python
# Convert 'Profit_Margin' to a 1D NumPy array and inspect its dimensions using .shape
profit_array = np.array(df["Profit_Margin"])
print("Array Shape:")
print(profit_array.shape)
```

```python
# Extract 'Profit_Margin' as a 1D Pandas Series and verify its row count via .shape
profit_series = df["Profit_Margin"]
print(profit_series.shape)
```

```python
# Subset 'Units_Sold' and 'Profit_Margin' into a 2D Pandas DataFrame and verify its dimensions
sales_df = df[["Units_Sold", "Profit_Margin"]]
print(sales_df.shape)
```
