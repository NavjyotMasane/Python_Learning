# Validate Data Using Regular Expressions

```python
# Validate PNR Number Format
# Import re module and write function to validate PNR number
```

```python
import re

def validate_pnr(pnr):
    pattern = r'^\d{8}$'
    if re.fullmatch(pattern, pnr):
        print("Valid PNR :", pnr)
    else:
        print("Invalid PNR :", pnr)
```

```python
# Test with sample PNR
validate_pnr("43821756")   # Valid
validate_pnr("6793452")    # Invalid - only 7 digits
validate_pnr("8120AB63")   # Invalid - contains letters
validate_pnr("81204763")   # Valid
```

```python
# Validate Seat Number Format
```

```python
# Write Seat Validator function
import re

def validate_seat(seat):
    pattern = r'^[A-Z][0-9]{1,2}$'
    if re.fullmatch(pattern, seat):
        print("Valid Seat Number :", seat)
    else:
        print("Invalid Seat Number :", seat)
```

```python
# Test with seat numbers
validate_seat("B4")    # Valid
validate_seat("A12")   # Valid
validate_seat("D23")   # Valid
validate_seat("b4")    # Invalid - lowercase letter
validate_seat("23B")   # Invalid - wrong order
validate_seat("A")     # Invalid - missing digit
```

```python
# Write a function to extract all train numbers
import re

def extract_train_numbers(text):
    pattern = r'\b[0-9]{4,5}\b'
    results = re.findall(pattern, text)
    print("Train Numbers Found :", results)
    return results
```

```python
# Combine all patterns into one complete ticket validator class
import re

class TicketValidator:

    def __init__(self, pnr, train_no, seat, fare):
        self.pnr      = pnr
        self.train_no = train_no
        self.seat     = seat
        self.fare     = fare

    def validate(self):
        print("=" * 40)
        print("   IndRail Ticket Validation Report")
        print("=" * 40)

        # Validate PNR - exactly 8 digits
        if re.fullmatch(r'[0-9]{8}', self.pnr):
            print("PNR      :", self.pnr, "-> VALID")
        else:
            print("PNR      :", self.pnr, "-> INVALID")

        # Validate Train Number - 4 or 5 digits
        if re.fullmatch(r'[0-9]{4,5}', self.train_no):
            print("Train No :", self.train_no, "-> VALID")
        else:
            print("Train No :", self.train_no, "-> INVALID")

```

```python
# Create ticket Objects
ticket1 = TicketValidator("43821756", "12951", "B4",  "880")
ticket2 = TicketValidator("6793452",  "1230X", "b12", "XX")

ticket1.validate()
ticket2.validate()
```

```python
# Create the final RailwayScraper class.
import requests
from bs4 import BeautifulSoup
 
class RailwayScraper:
 
    def __init__(self):
        # FIX: Replaced broken repo link with the official assignment platform URL
        self.base_url = "https://contentitvedant.github.io/Railway_Booking/"
 
    def get_soup(self, page):
        url = self.base_url + page
        try:
            response = requests.get(url)
            print("Fetched :", url, "| Status :", response.status_code)
            
            # If page fails to return an HTTP OK status, exit gracefully
            if response.status_code != 200:
                return None
                
            return BeautifulSoup(response.text, "html.parser")
        except requests.exceptions.RequestException as e:
            print(f"Network error trying to fetch {page}: {e}")
            return None
 
    def get_train_schedule(self):
        soup = self.get_soup("trains.html")
        if not soup:
            return []
            
        table = soup.find(id="train-schedule-table")
        # SAFEGUARD: Protect against attribute errors if id structure fails
        if not table or not table.find("tbody"):
            print("Error: Could not locate train schedule table structure.")
            return []
            
        rows = table.find("tbody").find_all("tr")
        trains = []
        for row in rows:
            cells = row.find_all("td")
            if len(cells) >= 7:
                trains.append({
                    "train_no"   : cells[0].get_text(strip=True),
                    "train_name" : cells[1].get_text(strip=True),
                    "from"       : cells[2].get_text(strip=True),
                    "to"         : cells[3].get_text(strip=True),
                    "departure"  : cells[4].get_text(strip=True),
                    "arrival"    : cells[5].get_text(strip=True),
                    "duration"   : cells[6].get_text(strip=True),
                })
        return trains
 
    def get_passengers(self):
        soup = self.get_soup("passengers.html")
        if not soup:
            return []
            
        table = soup.find(id="passenger-records-table")
        if not table or not table.find("tbody"):
            print("Error: Could not locate passenger table structure.")
            return []
            
        rows = table.find("tbody").find_all("tr")
        passengers = []
        for row in rows:
            # SAFEGUARD: Confirm the custom data attribute exists
            if not row.has_attr("data-pnr"):
                continue
                
            pnr = row["data-pnr"]
            name_el = row.find(class_="passenger-name")
            date_el = row.find(class_="journey-date")
            class_el = row.find(class_="travel-class")
            seat_el = row.find(class_="seat-number")
            fare_el = row.find(class_="fare")
            
            # Append only if child element properties match expected selectors
            if all([name_el, date_el, class_el, seat_el, fare_el]):
                passengers.append({
                    "pnr"  : pnr,
                    "name" : name_el.get_text(strip=True),
                    "date" : date_el.get_text(strip=True),
                    "class": class_el.get_text(strip=True),
                    "seat" : seat_el.get_text(strip=True),
                    "fare" : fare_el.get_text(strip=True),
                })
        return passengers
 
    def get_seat_availability(self):
        soup = self.get_soup("seat-availability.html")
        if not soup:
            return []
            
        table = soup.find(id="seat-availability-table")
        if not table or not table.find("tbody"):
            print("Error: Could not locate seat availability table structure.")
            return []
            
        rows = table.find("tbody").find_all("tr")
        available = []
        for row in rows:
            if not row.has_attr("data-train"):
                continue
                
            name_el = row.find(class_="train-name")
            sl_el = row.find(class_="seats-sl")
            a3_el = row.find(class_="seats-3a")
            a2_el = row.find(class_="seats-2a")
            a1_el = row.find(class_="seats-1a")
            
            if all([name_el, sl_el, a3_el, a2_el, a1_el]):
                available.append({
                    "train_no"   : row["data-train"],
                    "train_name" : name_el.get_text(strip=True),
                    "seats_sl"   : sl_el.get_text(strip=True),
                    "seats_3a"   : a3_el.get_text(strip=True),
                    "seats_2a"   : a2_el.get_text(strip=True),
                    "seats_1a"   : a1_el.get_text(strip=True),
                })
        return available

# Create Object and call all 3 Methods
# --- MAIN EXECUTION SCRIPT ---
scraper = RailwayScraper()

print("\n--- TRAIN SCHEDULE ---")
train_data = scraper.get_train_schedule()
for train in train_data:
    print(train["train_no"], "|", train["train_name"], "|", train["from"], "->", train["to"], "|", train["duration"])

print("\n--- PASSENGER RECORDS ---")
passenger_data = scraper.get_passengers()
for passenger in passenger_data:
    print("PNR:", passenger["pnr"], "|", "Name:", passenger["name"], "|", "Class:", passenger["class"], "|", "Seat:", passenger["seat"])
 
print("\n--- SEAT AVAILABILITY ---")
seat_data = scraper.get_seat_availability()
for seat in seat_data:
    print("Train", seat["train_no"], "-", seat["train_name"], "|", "SL:", seat["seats_sl"], "|", "3A:", seat["seats_3a"], "|", "2A:", seat["seats_2a"])
```
